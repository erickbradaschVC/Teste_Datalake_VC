```python
%sql
SELECT
    tp_movimento,
    COUNT(*) AS total,

    COUNT_IF(
        qt_movimento = 0
        AND num_pedido_compra IS NOT NULL
        AND TRIM(num_pedido_compra) <> ''
    ) AS zero_com_pedido,

    COUNT_IF(
        qt_movimento <> 0
        AND num_pedido_compra IS NOT NULL
        AND TRIM(num_pedido_compra) <> ''
    ) AS nao_zero_com_pedido,

    ROUND(
        100.0 * COUNT_IF(
            qt_movimento = 0
            AND num_pedido_compra IS NOT NULL
            AND TRIM(num_pedido_compra) <> ''
        ) / NULLIF(COUNT(*), 0),
        2
    ) AS pct_zero_com_pedido

FROM mb51_teste_4128
GROUP BY tp_movimento
ORDER BY total DESC;
```

```python
%sql
SELECT
    tp_movimento,
    desc_tp_movimento,
    COUNT(*) AS total_linhas,

    COUNT_IF(
        nm_usuario IS NULL
        OR LOWER(TRIM(nm_usuario)) IN (
            '',
            'null',
            'none',
            'nan',
            '#',
            '-'
        )
    ) AS usuario_vazio,

    COUNT_IF(
        nm_usuario IS NOT NULL
        AND LOWER(TRIM(nm_usuario)) NOT IN (
            '',
            'null',
            'none',
            'nan',
            '#',
            '-'
        )
    ) AS usuario_preenchido,

    ROUND(
        100.0 * COUNT_IF(
            nm_usuario IS NOT NULL
            AND LOWER(TRIM(nm_usuario)) NOT IN (
                '',
                'null',
                'none',
                'nan',
                '#',
                '-'
            )
        ) / NULLIF(COUNT(*), 0),
        4
    ) AS pct_preenchido,

    APPROX_COUNT_DISTINCT(
        CASE
            WHEN nm_usuario IS NOT NULL
             AND LOWER(TRIM(nm_usuario)) NOT IN (
                 '',
                 'null',
                 'none',
                 'nan',
                 '#',
                 '-'
             )
            THEN TRIM(nm_usuario)
        END
    ) AS usuarios_distintos

FROM mb51_teste_4128
GROUP BY
    tp_movimento,
    desc_tp_movimento
ORDER BY
    total_linhas DESC;
```

```python
%sql
SELECT
    tp_movimento,
    desc_tp_movimento,
    COUNT(*) AS total_linhas,

    COUNT_IF(
        tp_estoque_especial IS NULL
        OR LOWER(TRIM(tp_estoque_especial)) IN (
            '',
            'null',
            'none',
            'nan',
            '#',
            '-'
        )
    ) AS estoque_especial_vazio,

    COUNT_IF(
        tp_estoque_especial IS NOT NULL
        AND LOWER(TRIM(tp_estoque_especial)) NOT IN (
            '',
            'null',
            'none',
            'nan',
            '#',
            '-'
        )
    ) AS estoque_especial_preenchido,

    APPROX_COUNT_DISTINCT(
        CASE
            WHEN tp_estoque_especial IS NOT NULL
             AND LOWER(TRIM(tp_estoque_especial)) NOT IN (
                 '',
                 'null',
                 'none',
                 'nan',
                 '#',
                 '-'
             )
            THEN TRIM(tp_estoque_especial)
        END
    ) AS valores_distintos

FROM mb51_teste_4128
GROUP BY
    tp_movimento,
    desc_tp_movimento
ORDER BY
    total_linhas DESC;
```

```python
%sql
SELECT
    CASE
        WHEN tp_estoque_especial IS NULL
          OR LOWER(TRIM(tp_estoque_especial)) IN (
              '',
              'null',
              'none',
              'nan',
              '#',
              '-'
          )
        THEN '<VAZIO>'
        ELSE TRIM(tp_estoque_especial)
    END AS estoque_especial,

    COUNT(*) AS quantidade,

    ROUND(
        100.0 * COUNT(*) / SUM(COUNT(*)) OVER (),
        4
    ) AS percentual,

    COLLECT_SET(tp_movimento) AS movimentos

FROM mb51_teste_4128
GROUP BY
    CASE
        WHEN tp_estoque_especial IS NULL
          OR LOWER(TRIM(tp_estoque_especial)) IN (
              '',
              'null',
              'none',
              'nan',
              '#',
              '-'
          )
        THEN '<VAZIO>'
        ELSE TRIM(tp_estoque_especial)
    END
ORDER BY quantidade DESC;
```

```python
%sql
SELECT
    cod_material,
    COUNT(DISTINCT sg_unidade_medida) AS unidades_distintas,
    COLLECT_SET(sg_unidade_medida) AS unidades,
    COUNT(*) AS movimentos
FROM mb51_teste_4128
WHERE cod_material IS NOT NULL
  AND TRIM(cod_material) <> ''
GROUP BY cod_material
HAVING COUNT(DISTINCT sg_unidade_medida) > 1
ORDER BY
    unidades_distintas DESC,
    movimentos DESC;
```

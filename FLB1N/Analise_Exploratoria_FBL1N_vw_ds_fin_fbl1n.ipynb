## 4. Granularidade e chave real

`linhas ÷ chaves distintas`. Razão maior que 1,00 indica dimensão adicional multiplicando linhas.

```sql
-- GRANULARIDADE
WITH t AS (SELECT COUNT(*) AS total FROM dev_procurement.corp_curated.vw_ds_fin_fbl1n),
g AS (
  SELECT 'num_lancamento_contabil' AS chave, COUNT(*) AS distintos FROM (SELECT DISTINCT `num_lancamento_contabil` FROM dev_procurement.corp_curated.vw_ds_fin_fbl1n)
UNION ALL
  SELECT 'num_lancamento_contabil + num_item_lancamento_contabil' AS chave, COUNT(*) AS distintos FROM (SELECT DISTINCT `num_lancamento_contabil`, `num_item_lancamento_contabil` FROM dev_procurement.corp_curated.vw_ds_fin_fbl1n)
UNION ALL
  SELECT 'cod_empresa + num_exercicio_fiscal + num_lancamento_contabil' AS chave, COUNT(*) AS distintos FROM (SELECT DISTINCT `cod_empresa`, `num_exercicio_fiscal`, `num_lancamento_contabil` FROM dev_procurement.corp_curated.vw_ds_fin_fbl1n)
UNION ALL
  SELECT 'cod_empresa + num_exercicio_fiscal + num_lancamento_contabil + num_item_lancamento_contabil' AS chave, COUNT(*) AS distintos FROM (SELECT DISTINCT `cod_empresa`, `num_exercicio_fiscal`, `num_lancamento_contabil`, `num_item_lancamento_contabil` FROM dev_procurement.corp_curated.vw_ds_fin_fbl1n)
)
SELECT g.chave, t.total AS linhas, g.distintos,
       ROUND(t.total / g.distintos, 4) AS linhas_por_chave,
       CASE WHEN g.distintos = t.total THEN 'CHAVE UNICA'
            ELSE 'NAO UNICA - ha dimensao adicional' END AS veredito
FROM g CROSS JOIN t
ORDER BY linhas_por_chave;
```

## 18. Análises específicas — FBL1N

### 18.1 Pendência declarada em `vl_moeda_interna`

O comentário do schema registra textualmente uma **pendência em aberto**: o dicionário de campos
aponta `DMSHB` (montante em moeda da **empresa**), mas a fonte atual seria a moeda da **transação**.

Esta célula compara as duas colunas. Se forem idênticas em toda a base, a pendência se confirma:
`vl_moeda_interna` está replicando `vl_moeda_transacao` em vez de trazer o valor em moeda interna.

```sql
-- 18.1 vl_moeda_interna vs vl_moeda_transacao
SELECT COUNT(*)                                                          AS linhas,
       COUNT_IF(vl_moeda_interna = vl_moeda_transacao)                   AS valores_identicos,
       COUNT_IF(vl_moeda_interna <> vl_moeda_transacao)                  AS valores_diferentes,
       COUNT_IF(vl_moeda_interna IS NULL)                                AS interna_nula,
       COUNT_IF(vl_moeda_transacao IS NULL)                              AS transacao_nula,
       ROUND(SUM(vl_moeda_interna), 2)                                   AS soma_interna,
       ROUND(SUM(vl_moeda_transacao), 2)                                 AS soma_transacao,
       CASE WHEN COUNT_IF(vl_moeda_interna <> vl_moeda_transacao) = 0
            THEN 'PENDENCIA CONFIRMADA - colunas identicas'
            ELSE 'colunas divergem - investigar a regra' END             AS veredito
FROM dev_procurement.corp_curated.vw_ds_fin_fbl1n;
```

### 18.2 Coerência entre moeda e valor

A tabela tem 3 colunas de código de moeda e 6 de valor. Esta célula verifica se os valores
em moeda adicional só aparecem quando a moeda correspondente está preenchida.

```sql
-- 18.2 MOEDA x VALOR
SELECT 'moeda_transacao vazia MAS vl_moeda_transacao preenchido' AS situacao,
       COUNT_IF((cod_moeda_transacao IS NULL OR trim(cod_moeda_transacao) = '')
            AND vl_moeda_transacao IS NOT NULL AND vl_moeda_transacao <> 0) AS linhas
  FROM dev_procurement.corp_curated.vw_ds_fin_fbl1n
UNION ALL
SELECT 'moeda_adicional_2 vazia MAS vl_moeda_adicional_2 preenchido',
       COUNT_IF((cod_moeda_adicional_2 IS NULL OR trim(cod_moeda_adicional_2) = '')
            AND vl_moeda_adicional_2 IS NOT NULL AND vl_moeda_adicional_2 <> 0)
  FROM dev_procurement.corp_curated.vw_ds_fin_fbl1n
UNION ALL
SELECT 'desconto preenchido MAS base_desconto zerada',
       COUNT_IF(vl_desconto <> 0 AND (vl_base_desconto IS NULL OR vl_base_desconto = 0))
  FROM dev_procurement.corp_curated.vw_ds_fin_fbl1n
ORDER BY linhas DESC;
```

### 18.3 Ciclo de compensação

Partida compensada deve ter data de compensação e documento de compensação preenchidos.
Combinações inconsistentes indicam problema de carga.

```sql
-- 18.3 CICLO DE COMPENSACAO
SELECT COUNT(*)                                                             AS linhas,
       COUNT_IF(dt_compensacao IS NOT NULL AND trim(dt_compensacao) NOT IN ('', '00000000')) AS com_data_compensacao,
       COUNT_IF(cod_lancamento_contabil_compensacao IS NOT NULL
            AND trim(cod_lancamento_contabil_compensacao) <> '')            AS com_doc_compensacao,
       COUNT_IF((dt_compensacao IS NULL OR trim(dt_compensacao) IN ('', '00000000'))
            AND cod_lancamento_contabil_compensacao IS NOT NULL
            AND trim(cod_lancamento_contabil_compensacao) <> '')            AS doc_sem_data,
       COUNT_IF(dt_compensacao IS NOT NULL AND trim(dt_compensacao) NOT IN ('', '00000000')
            AND (cod_lancamento_contabil_compensacao IS NULL
              OR trim(cod_lancamento_contabil_compensacao) = ''))           AS data_sem_doc,
       COUNT_IF(ind_compensacao_estornada IS NOT NULL
            AND trim(ind_compensacao_estornada) <> '')                      AS compensacao_estornada
FROM dev_procurement.corp_curated.vw_ds_fin_fbl1n;
```

### 18.4 Formato de `cod_exercicio_periodo`

O comentário declara o formato `aaaa/pp` (exercício e período concatenados).
Esta célula verifica se o formato é consistente e se bate com as colunas atômicas
`num_exercicio_fiscal` e `cod_periodo_contabil`.

```sql
-- 18.4 FORMATO E COERENCIA DE cod_exercicio_periodo
SELECT COUNT(*)                                                          AS linhas,
       COUNT_IF(trim(cod_exercicio_periodo) RLIKE '^[0-9]{4}/[0-9]{2}$') AS formato_aaaa_pp,
       COUNT_IF(NOT trim(cod_exercicio_periodo) RLIKE '^[0-9]{4}/[0-9]{2}$'
            AND trim(cod_exercicio_periodo) <> '')                        AS fora_do_formato,
       COUNT_IF(substring(trim(cod_exercicio_periodo), 1, 4) = trim(num_exercicio_fiscal)) AS exercicio_bate,
       COUNT_IF(substring(trim(cod_exercicio_periodo), 1, 4) <> trim(num_exercicio_fiscal)) AS exercicio_diverge
FROM dev_procurement.corp_curated.vw_ds_fin_fbl1n;
```

```sql
-- 20. RESUMO
WITH t AS (SELECT COUNT(*) AS total FROM dev_procurement.corp_curated.vw_ds_fin_fbl1n),
g AS (
  SELECT 'num_lancamento_contabil' AS chave, COUNT(*) AS d FROM (SELECT DISTINCT `num_lancamento_contabil` FROM dev_procurement.corp_curated.vw_ds_fin_fbl1n)
UNION ALL
  SELECT 'num_lancamento_contabil + num_item_lancamento_contabil' AS chave, COUNT(*) AS d FROM (SELECT DISTINCT `num_lancamento_contabil`, `num_item_lancamento_contabil` FROM dev_procurement.corp_curated.vw_ds_fin_fbl1n)
UNION ALL
  SELECT 'cod_empresa + num_exercicio_fiscal + num_lancamento_contabil' AS chave, COUNT(*) AS d FROM (SELECT DISTINCT `cod_empresa`, `num_exercicio_fiscal`, `num_lancamento_contabil` FROM dev_procurement.corp_curated.vw_ds_fin_fbl1n)
UNION ALL
  SELECT 'cod_empresa + num_exercicio_fiscal + num_lancamento_contabil + num_item_lancamento_contabil' AS chave, COUNT(*) AS d FROM (SELECT DISTINCT `cod_empresa`, `num_exercicio_fiscal`, `num_lancamento_contabil`, `num_item_lancamento_contabil` FROM dev_procurement.corp_curated.vw_ds_fin_fbl1n)
)
SELECT 'IDENTIFICACAO' AS bloco, 'transacao' AS item, 'FBL1N' AS valor
UNION ALL SELECT 'IDENTIFICACAO', 'tabela', 'vw_ds_fin_fbl1n'
UNION ALL SELECT 'IDENTIFICACAO', 'tipo de objeto', 'VIEW'
UNION ALL SELECT 'IDENTIFICACAO', 'linhas', format_number((SELECT total FROM t), 0)
UNION ALL SELECT 'IDENTIFICACAO', 'colunas', '73'
UNION ALL SELECT 'IDENTIFICACAO', 'coluna de recorte', 'cod_centro'
UNION ALL
SELECT 'GRANULARIDADE', g.chave,
       CONCAT(format_number(g.d, 0), ' distintos | ',
              CAST(ROUND(t.total / g.d, 4) AS STRING), ' linhas/chave | ',
              CASE WHEN g.d = t.total THEN 'CHAVE UNICA' ELSE 'nao unica' END)
  FROM g CROSS JOIN t
UNION ALL
SELECT 'CHAVE REAL', 'sugerida',
       COALESCE((SELECT MIN(g.chave) FROM g CROSS JOIN t WHERE g.d = t.total),
                'NENHUMA - investigar')
ORDER BY bloco, item;
```
